```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Chart 1 - Companies per sector
df['Sector'].value_counts().plot(kind='bar', ax=axes[0], color='steelblue')
axes[0].set_title('Companies per Sector')
axes[0].set_xlabel('Sector')
axes[0].set_ylabel('Number of Companies')
axes[0].tick_params(axis='x', rotation=45)

# Chart 2 - Top 10 companies by market cap
top10 = df[['Name', 'Market Cap']].sort_values('Market Cap', ascending=False).head(10)
axes[1].barh(top10['Name'], top10['Market Cap'])
axes[1].set_title('Top 10 Companies by Market Cap')
axes[1].set_xlabel('Market Cap')

# Chart 3 - Average P/E ratio by sector
df.groupby('Sector')['Price/Earnings'].mean().sort_values().plot(kind='barh', ax=axes[2], color='green')
axes[2].set_title('Average P/E Ratio by Sector')
axes[2].set_xlabel('P/E Ratio')

plt.tight_layout()
plt.show()
```

```python
# Finding undervalued sectors - high EBITDA, low P/E
sector_summary = df.groupby('Sector').agg({
    'Price/Earnings': 'mean',
    'Market Cap': 'mean',
    'Dividend Yield': 'mean',
    'EBITDA': 'mean'
}).round(2)

sector_summary.sort_values('Price/Earnings')
```

```python
pd.set_option('display.float_format', lambda x: '{:,.2f}'.format(x))

sector_summary.sort_values('Price/Earnings')
```
